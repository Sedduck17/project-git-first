##3.3 Preprocessing Data

```python
import pandas as pd

data = {
"nama": ["Andi", "Budi", "Citra", "Dini", "Eka", "Fajar", "Fajar"],
"jam_belajar": [5, 2, 8, None, 6, 3, 3],
"jumlah_latihan": [10, 3, 15, 2, None, 5, 5],
"kehadiran": [90, 60, 95, 50, 85, 70, 70],
"gender": ["Laki-laki", "Laki-laki", "Perempuan", "Perempuan",
"Perempuan", "Laki-laki", "Laki-laki"],
"lulus": ["Ya", "Tidak", "Ya", "Tidak", "Ya", "Tidak", "Tidak"]
}

df = pd.DataFrame(data)

print(df)
```

```python
#cek missing value

df.isna().sum()
```

```python
#Mengisi nilai kosong
rata_jam = df["jam_belajar"].mean()
df['jam_belajar'] = df["jam_belajar"].fillna(rata_jam)

rata_jam = df["jumlah_latihan"].mean()
df["jumlah_latihan"] = df["jumlah_latihan"].fillna(rata_jam).round(1)
print(df)
```

```python
#Cek Duplikat
df.duplicated().sum()

df = df.drop_duplicates()
print(df)
```

```python
#Cek Tipe Data
df.info()
df["jumlah_latihan"] = df['jumlah_latihan'].astype(int)
print(df)
```

```python
df['lulus_encode'] = df["lulus"].map({
    "Tidak" : 0,
    "Ya" : 1
})
print(df)
```

```python
gender_dummies = pd.get_dummies(df['gender'])

print(gender_dummies)
```

```python
#scale data numerik

df['jam_belajar_scale'] = (df["jam_belajar"] - df["jam_belajar"].min()) / (df["jam_belajar"].max() - df["jam_belajar"].min())
df['jam_belajar_scale'] = df['jam_belajar_scale'].round(1)
print(df)
```

```python
#data outlier
df.describe()
```

##Klasifikasi dengan logistik regression

```python
import pandas as pd

df = pd.read_csv("/content/Iris.csv")

df.head()
```

```python
x = df[["SepalLengthCm", "SepalWidthCm", "PetalLengthCm" , "PetalWidthCm"]]
y = df["Species"]

print("feature : ", x)
print("\ntarget", y)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2 ,
    random_state=42,
    stratify = y
)

print("ukuran x_train", x_train.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
model_logistic = LogisticRegression(max_iter=200)

model_logistic.fit(x_train, y_train)
prediksi_logistic = model_logistic.predict(x_test)
accuracy_logistic = accuracy_score(y_test, prediksi_logistic)

print("Hasil prediksi : ")
print(prediksi_logistic)

print("Akurasi : ")
print(accuracy_logistic)
```

```python
probabilitas = model_logistic.predict_proba(x_test)

print(probabilitas[:5])
```
